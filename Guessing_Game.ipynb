```python
jackpot=random.randint(1,100)
guess=int(input("Guss the number"))
counter=1
while guess != jackpot:
    if guess < jackpot:
        print("Guess Higher")
    else:
        print("Guess Lower")
    guess=int(input("Guss the number"))
    counter+=1

print("Right Answer")
print("You took",counter,"attempts")
```
